```
import numpy as np
import matplotlib.pyplot as plt
```

```
x = np.linspace(0, 100, 30)
```

```
x
```

```
y = x**2
```

```
y
```

```
plt.plot(x,y, "g-*")
plt.xlabel("Valores Eixo X")
plt.ylabel("Valores Eixo Y")
plt.title("Exemplo de gráfico")
```

```
plt.bar(x,y, color="r")
plt.xlabel("Valores Eixo X")
plt.ylabel("Valores Eixo Y")
plt.title("Exemplo de gráfico")
```

```
plt.subplot(1, 2, 1)
```

```
plt.subplot(1, 2, 1)
plt.plot(x,y, "g-*")
plt.xlabel("Valores Eixo X")
plt.ylabel("Valores Eixo Y")
plt.title("Exemplo de gráfico")
plt.subplot(1, 2, 2)
plt.bar(x,x, color = "y")
plt.xlabel("Valores Eixo X")
plt.ylabel("Valores Eixo Y")
plt.title("Exemplo de gráfico")
```

```
plt.figure(figsize=(12,5))
plt.subplot(1, 2, 1)
plt.plot(x, y, "r--")
plt.subplot(1, 2, 2)
plt.bar(x, 2*x, color="green")
```

```
fig = plt.figure()
axes1 = fig.add_axes([0.1, 0.1, 1, 1])
axes2 = fig.add_axes([0.6, 0.8, 0.2, 0.2])

axes1.plot(x, y, "r--")
axes2.bar(x, y**2, color="green")
```

```
import numpy as np
import matplotlib.pyplot as plt
```

```
x = np.linspace(0, 100, 30)
y = x**2
```

```
fig, ax = plt.subplots(2,2)
```

```
fig, ax = plt.subplots(2,2, figsize=(10,7))
ax[0,0].plot(x,y, "r--")
ax[0,0].set_xlabel("Eixo X primeiro graf")
ax[0,1].plot(x,y, "g--")
ax[1,0].plot(x,y**2, "b*-")
ax[1,1].bar(x,y)

plt.tight_layout()
```
