```
import pandas as pd
import matplotlib.pyplot as plt
```

```
df = pd.read_csv('Salaries.csv')
```

```
df.head()
```

```
df['Year']
```

```
df = df[df["BasePay"] != "Not Provided"]
```

```
df["BasePay"] = df["BasePay"].astype(float)
```

```
df_plots = df[["BasePay", "Year", "JobTitle"]]
```

```
df_plots.info()
```

```
df_plots.groupby("Year").size()
```

```
df_plots["Year"] = df["Year"].astype(str)
qtd_ano = df_plots.groupby("Year").size()
```

```
plt.figure(figsize = (8,5))

plt.bar(qtd_ano.index, qtd_ano.values)
```

```
plt.figure(figsize = (8,5))

plt.bar(qtd_ano.index, qtd_ano.values, color="lightgreen")
plt.title("Quantidade por Ano")
plt.xlabel("Ano")
plt.ylabel("Quantidade")
```

```
media_salario = df_plots.groupby("Year")["BasePay"].mean()
```

```
plt.figure(figsize = (8,5))

plt.plot(media_salario.index, media_salario.values, "b*-")
plt.title("Média Salarial por Ano")
plt.xlabel("Ano")
plt.ylabel("Quantidade")
```

```
fig, ax = plt.subplots(nrows=1, ncols=2, figsize = (10,5))

#Gráfico de barras
ax[0].bar(qtd_ano.index, qtd_ano.values, color="lightgreen")
ax[0].set_title("Quantidade por Ano")
ax[0].set_xlabel("Ano")
ax[0].set_ylabel("Quantidade")

#Gráfico de linha
ax[1].plot(media_salario.index, media_salario.values, "b*-")
ax[1].set_title("Média Salarial por Ano")
ax[1].set_xlabel("Ano")
ax[1].set_ylabel("Média")

plt.tight_layout()
```
